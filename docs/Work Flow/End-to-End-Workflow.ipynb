### Phase 1 – Requirements & Use Case Definition
| # | Step | Role | Activity & Output |
|---|---|---|---|
| 1 | Submit Request | End User | States the business question and scope (e.g., BMI statistics by wave, race, gender, age, cohort). |
| 2 | Analysis & Data Requirements | Data Analyst | Identifies RAND variables needed (roots, labels, Codebook pages), checks existing Silver CDM coverage, and defines grain, measures, and cohort scope. |
| 3 | Create Business Use Case Document | Data Analyst | Documents use case name, description, required variables, expected grain, cohort scope, and wave range. Serves as the business authorization; the agent never reads the document. |
| 4 | Review & Approve | End User | Reviews and signs off on the use case document. No downstream process should proceed without approval. |

### Phase 2 – Metadata Foundation
| # | Step | Role | Activity & Output |
|---|---|---|---|
| 5 | Foundation Notebooks (1.1–1.4.2) | Data Engineer | Runs table extraction, Codebook page processing (2,133 pages), variable candidate generation, review tables, and bronze-column matching. Output includes hrs_document_pages and the candidate/review chain. |

### Phase 3 – Variable Dictionary, Approval & Coverage
| # | Step | Role | Activity & Output |
|---|---|---|---|
| 6 | Curate Dictionary (1.5.1) | Data Analyst | Creates a use-case-specific dictionary containing only required variables. Verifies labels against Codebook pages and confirms bronze-column matches. |
| 7 | Approval Processing (1.5.2) | Data Analyst | Uses widgets use_case_name and codebook_version to generate hrs_use_case_approved_variables. Verifies expected variable counts and exact bronze matches. |
| 8 | Cohort Coverage (1.5.3) | Data Analyst | Uses the same widgets to generate hrs_use_case_approved_cohort_coverage. Reviews HAS_DATA and NO_DATA statuses. |
| 9 | Technical Sign-Off (Stage 4) | Data Analyst | Records reviewed_by, reviewed_at, and wave_coverage_status in approved tables after validating results against the Business Use Case Document. |
| 10 | Coverage Confirmation | End User | Optionally confirms that NO_DATA coverage implications are acceptable. |

### Phase 4 – Agent Drafting (Review Only)
| # | Step | Role | Activity & Output |
|---|---|---|---|
| 11 | Configure & Run Agent | Data Engineer | Sets HRS_AGENT_USE_CASE environment variable and executes the agent from src/agent. |
| 12 | Routing & Drafting | Agent | Produces coverage reports and drafts DDL/DML artifacts for gaps, including fact_{use_case_stem} objects and Codebook label dimensions. Agent is strictly review-only and never executes changes. |
| 13 | Review Drafts | Data Engineer | Reviews generated artifacts against type standards, grain requirements, and the review checklist. |

### Phase 5 – Deploy & Verify
| # | Step | Role | Activity & Output |
|---|---|---|---|
| 14 | Apply Drafts | Data Engineer | Moves approved DDL/DML into repository scripts and completes Git workflow (feature branch → PR → CI → merge). |
| 15 | Execute DDL + DML | Data Engineer | Executes scripts from repository files against dev_catalog. DDL is executed before DML. |
| 16 | Re-Run Agent Routing | Agent | Generates a coverage report confirming all approved variables route to their designated CDM homes. |
| 17 | Verify Loaded Tables | Data Engineer | Validates row counts, grain consistency, population rates, value ranges, foreign key integrity, and validation scripts. |
| 18 | Consume Results | End User | Consumes gold facts, dashboards, and reports. Feedback may generate new requirements and restart the workflow. |

### Role Boundaries
| Role | Owns | Never Does |
|--------|--------|--------|
| End User | Requirements, business approval of the use case document, coverage confirmation, and consuming results | N/A |
| Data Analyst | Business Use Case Document, dictionary curation, approval processing, coverage review, and technical sign-off | Executes DDL/DML or commits code |
| Data Engineer | Draft review, Git workflow, DDL/DML execution, and verification | Approves business requirements |
| Agent | Scoping, routing, coverage reporting, and DDL/DML draft generation based only on approved metadata | Executes SQL, writes to the warehouse, commits code, or reads the Business Use Case Document |
