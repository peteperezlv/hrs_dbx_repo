# Introduction
The HRS Silver CDM project provides a standardized data architecture and development process for transforming the **RAND HRS Longitudinal Dataset** into a structured Silver-layer Common Data Model (CDM) in Databricks.

The project is designed to make it easier to:

* Ingest RAND HRS data.
* Understand the HRS codebook and source variables.
* Organize the data into logical subject areas.
* Create standardized Delta tables.
* Transform the wide longitudinal RAND HRS structure into a respondent-by-wave structure.
* Support SQL-based analytics.
* Support dashboards and data visualizations.
* Add new HRS subject areas using a repeatable process.

The initial subject areas may include:

* Demographics
* Health
* Functional Limitations and Helpers
* Financial and Housing Wealth
* Income
* Social Security
* Pensions
* Health Insurance
* Family Structure
* Retirement Plans and Expectations
* Employment History
* Psycho-Social / Leave-Behind

The architecture is designed so that additional HRS subject areas can be added without redesigning the overall framework.

# Business Objective

Analyze HRS Health and Retirement questions that need to be answered with data visualizations and statistical reports. 

Enable researchers, students, and analysts to use the RAND Health and Retirement Study (HRS) Longitudinal Dataset to perform descriptive statistical analysis on various measures, for a selected sample population. The analysis will provide insights into the central tendency and variability metrics of HRS health measures and support comparisons across various dimension such as wave, cohort, race, ethnicity, etc.

# Core Definitions
- Measures: The raw, fundamental data points or simple numeric values with no added context (e.g., total dollar amount of a single sale, or a count of 500 website visits).

- Metrics: A standard calculation or aggregation of one or more measures that provides operational context, such as an average, ratio, or rate (e.g., Average Order Value calculated by total revenue divided by the number of orders).

- KPIs (Key Performance Indicators): A high-level, critical subset of metrics explicitly tied to a specific business objective, target, and timeframe (e.g., reaching $50,000 in monthly recurring revenue by the end of the quarter).

# Business Questions

The following sections provide a list of business questions provided by the the user.

## Q5 Group Questions

1. Q5.1 Residents of a retirement home were asked how many of the previous years they had received the flu vaccine. Determine the mode and median from the residents who provided answers.
1. Q5.2 Which measure of central tendency (mean, median or mode) best represents the number of previous years that the residents of this retirement home received the flu vaccine? Why?
1. Q5.3 How do BMI statistics vary by wave, race, gender, age and HRS cohort?
1. Q5.4 What is the mean age of the sample and what is the median age? 
1. Q5.5 What is the mean BMI?  What is the mode for BMI?
1. Q5.6 What does it mean when the mean and median are similar or the same like in this sample for age? What does the mode tell you about the age of the sample? 
1. Q5.7 Which race/ethnic group has the highest average BMI? Which has the lowest? What race/ethnic group reports the highest maximum BMI in our sample? What was the max BMI reported?
1. Q5.8 Which race/ethnic group has the greatest variability (or the most dispersion) in the sample? How can you tell? What does that mean?
