# 2. Purpose

This document provides a step-by-step process for creating a **Gold Layer DDL Functional Specification** for an HRS analytical use case.

The workflow helps the user move from an analytical requirement to a completed DDL specification that can be used to generate a Databricks SQL DDL script.

The workflow is designed to ensure that:

* The analytical use case is clearly defined.
* The appropriate RAND HRS variables are identified.
* The Gold table grain is understood.
* The physical table structure is clearly defined.
* Keys and relationships are identified.
* The DDL specification is reviewed before SQL is generated.
* The resulting Delta table can be validated before beginning DML development.

---

# 3. DDL Workflow Overview

The overall process is:

```text
Document the Use Case
        │
        ▼
Obtain the RAND HRS Codebook
        │
        ▼
Identify Variables Supporting the Use Case
        │
        ▼
Create a New DDL Specification
        │
        ▼
Complete the Gold DDL Template
        │
        ▼
Review the DDL Specification
        │
        ▼
Generate the DDL SQL
        │
        ▼
Review the Generated SQL
        │
        ▼
Execute DDL in Databricks
        │
        ▼
Validate the Gold Delta Table
        │
        ▼
Approved Gold Table
        │
        ▼
Begin Gold DML Specification
```

# 4. Step 1 — Document the Analytical Use Case

Before creating a Gold table, clearly document **why the table is needed**.

The use case should identify the analytical question that the Gold table will support.

### Document the following:

| Item              | Description                                   |
| ----------------- | --------------------------------------------- |
| Use Case ID       | Unique identifier for the analytical use case |
| Use Case Name     | Short descriptive name                        |
| Business Question | Question the analysis is intended to answer   |
| Primary Users     | Analysts, researchers, policy analysts, etc.  |
| Intended Use      | Analytics, visualization, reporting, etc.     |

### Example

```text
Use Case ID:
UC_DA_001

Use Case Name:
HRS BMI Race and Gender Analysis

Business Question:
How do BMI statistics vary across HRS cohorts,
survey waves, race groups, and gender groups?

Primary Users:
Data Analysts / Researchers / Policy Analysts

Intended Use:
Analytics / Visualization / Reporting
```

### Step 1 Deliverable

A clearly documented analytical use case.

---

# 5. Step 2 — Obtain the RAND HRS Codebook

The RAND HRS Codebook is a critical input to the Gold DDL design process.

The codebook provides information needed to understand the HRS variables, including variable names, descriptions, survey waves, and variable characteristics.

Obtain the appropriate RAND HRS Codebook corresponding to the HRS dataset being used.
* https://hrsdata.isr.umich.edu/data-products/rand-hrs-longitudinal-file-2022

### Codebook Requirements

Confirm that the codebook:

* Corresponds to the HRS dataset version being used.
* Contains the variables required for the use case.
* Provides the information needed to understand the variables.
* Is available for reference during specification development.

### Important

The DDL specification does not copy the entire codebook.

The codebook is used as a **reference source** for identifying and understanding the variables required by the analytical use case.

### Step 2 Deliverable

A copy of the applicable RAND HRS Codebook available for reference.

---

# 6. Step 3 — Identify Variables Supporting the Use Case

The **Use Case** and the **RAND HRS Codebook** are the two primary inputs to this step.

```text
        Use Case
           │
           │
           ▼
   Business Question
           │
           │
           ├───────────────┐
           │               │
           ▼               ▼
     HRS Variables    RAND HRS Codebook
           │               │
           └───────┬───────┘
                   ▼
          Identify Required
              Variables
```

The objective is to identify the HRS data elements required to support the analytical question.

### Identify:

* Dimensions
* Measures
* Identifiers
* Cohort information
* Wave information
* Demographic attributes
* Other analytical attributes required by the use case

### Example

For a BMI analysis, the required analytical elements might include:

```text
Cohort
Wave
Race
Gender
BMI
```

### Important DDL/DML Distinction

At this stage, identify the **business data elements** needed for the analysis.

Do not yet design the transformation or loading SQL.

The eventual Gold DML specification will define:

* Source tables
* Source variables
* Transformations
* Calculations
* Aggregations
* Loading logic

### Step 3 Deliverable

A documented list of the HRS variables and analytical elements required to support the use case.

---

# 7. Step 4 — Create the Gold DDL Specification

Create a new specification document.

There are two options.

### Option A — Start from the Master Template

Create a copy of:

```text
HRS_Gold_Layer_DDL_Master_Functional_Specification
```

Use this option when creating a new type of Gold table.

### Option B — Start from an Existing Specification

Copy an existing Gold DDL specification that is structurally similar to the new use case.

Use this option when creating another table with a similar design.

### Important

When copying an existing specification:

> Review every user-input section and remove or replace information belonging to the previous use case.

Do not assume that an existing specification is automatically correct for the new table.

### Step 4 Deliverable

A new working copy of the Gold DDL specification.

---

# 8. Step 5 — Complete the Specification

Complete the 🔵 **USER INPUT** sections of the Gold DDL template.

Review the 🟢 **CONFIRM** sections.

Do not modify 🟡 **MASTER CONTROLLED** sections unless the HRS architecture has changed.

### Complete the following areas:

#### Analytical Definition

* Use Case ID
* Use Case Name
* Business Question
* Primary Users
* Intended Use

#### Gold Table Design

* Gold Table Grain
* Gold Table Type
* Target Table Name
* Target Table Description

#### Physical Structure

* Dimensions
* Measures
* Analytical attributes
* Keys
* Foreign keys
* Constraints
* Audit columns
* Table comments
* Column comments

---

# 9. Step 6 — Define the Gold Table Grain

The table grain is one of the most important decisions in the specification.

Explicitly define:

> **What does one row represent?**

### Example

For BMI statistics:

```text
One row represents one combination of:

HRS Cohort
+
HRS Wave
+
Race
+
Gender
```

This means that BMI itself is **not** part of the grouping grain.

### Why Grain Matters

The grain determines:

* What constitutes one analytical observation.
* Which columns identify that observation.
* Which columns are dimensions.
* Which values are measures.
* Whether a uniqueness rule is appropriate.

### Step 6 Deliverable

A clearly documented Gold table grain.

---

# 10. Step 7 — Define the Physical Columns

Using the use case and identified analytical requirements, define the physical columns required in the Gold table.

Each column should specify:

| Attribute            | Required |
| -------------------- | -------- |
| Column Name          | Yes      |
| Category             | Yes      |
| Databricks Data Type | Yes      |
| Nullable             | Yes      |
| Description          | Yes      |

### Example

| Column    | Category    | Type   | Nullable | Description           |
| --------- | ----------- | ------ | -------- | --------------------- |
| cohort_id | Foreign Key | BIGINT | No       | HRS cohort identifier |
| wave_id   | Foreign Key | BIGINT | No       | HRS wave identifier   |
| raracem   | Dimension   | INT    | Yes      | Race category         |
| ragender  | Dimension   | INT    | Yes      | Gender category       |
| bmi_mean  | Measure     | DOUBLE | Yes      | Mean BMI              |

### Important

Only include columns required by the analytical design.

Do not add columns simply because they exist in the source dataset.

---

# 11. Step 8 — Define Keys and Relationships

Identify the keys required by the Gold table.

Consider:

* Surrogate primary key
* Analytical key
* Business key
* Foreign keys
* Uniqueness requirements

### Example

```text
Surrogate Primary Key:
fact_hrs_bmi_race_gender_stats_id

Foreign Keys:
cohort_id → dim_hrs_cohort.cohort_id
wave_id   → dim_hrs_wave.wave_id

Analytical Grain:
cohort_id + wave_id + raracem + ragender
```

### Important

The DDL specification defines the **physical relationships**.

The DML specification will later define how those key values are obtained and populated.

---

# 12. Step 9 — Review the Completed Specification

Before generating SQL, conduct a specification walkthrough.

The review should confirm that the specification accurately represents the analytical requirement.

### Review Questions

#### Analytical Design

* Is the business question clear?
* Is the intended analytical purpose clear?
* Is the table grain unambiguous?

#### Table Structure

* Are all required dimensions present?
* Are all required measures present?
* Are unnecessary columns excluded?
* Are the data types appropriate?
* Is nullability correct?

#### Keys

* Is the primary key clearly defined?
* Are foreign keys correctly identified?
* Are uniqueness requirements appropriate?

#### Documentation

* Does the table have a meaningful description?
* Does every column have a meaningful comment?

### Step 9 Deliverable

An approved Gold DDL Functional Specification.

---

# 13. Step 10 — Generate the DDL SQL

Submit the completed Gold DDL specification to the DDL generation process.

The AI assistant should generate a Databricks SQL DDL script based **only on the approved specification**.

The generated SQL should:

* Create the specified Gold table.
* Use the specified catalog and schema.
* Use Delta storage.
* Create the specified columns.
* Apply specified data types.
* Apply specified constraints.
* Create required foreign keys.
* Create required identity columns.
* Include table comments.
* Include column comments.

### The generated script must not contain:

```text
INSERT
UPDATE
DELETE
MERGE
SELECT
```

The output is **DDL only**.

### Step 10 Deliverable

A generated Databricks SQL DDL script.

---

# 14. Step 11 — Review the Generated SQL

Do not execute generated SQL without reviewing it.

Compare the SQL against the approved DDL specification.

### Verify:

* Catalog
* Schema
* Table name
* Columns
* Data types
* Nullability
* Primary key
* Foreign keys
* Constraints
* Identity column
* Audit columns
* Table comment
* Column comments

### Important

The generated SQL should not contain structural elements that were not defined in the specification.

---

# 15. Step 12 — Execute the DDL in Databricks

Execute the reviewed DDL in the appropriate Databricks development environment.

The expected result is a Gold Delta Managed Table.

### Expected Target

```text
dev_catalog.gld_star_hrs.<target_table>
```

### Step 12 Deliverable

A physically created Gold Delta table.

---

# 16. Step 13 — Validate the Gold Table

After executing the DDL, validate the physical table against the specification.

### Validate:

| Validation         | Required |
| ------------------ | -------- |
| Table Exists       | Yes      |
| Correct Catalog    | Yes      |
| Correct Schema     | Yes      |
| Correct Table Name | Yes      |
| Delta Format       | Yes      |
| Managed Table      | Yes      |
| Expected Columns   | Yes      |
| Data Types         | Yes      |
| Nullability        | Yes      |
| Identity Column    | Yes      |
| Primary Key        | Yes      |
| Foreign Keys       | Yes      |
| Constraints        | Yes      |
| Audit Columns      | Yes      |
| Table Comment      | Yes      |
| Column Comments    | Yes      |

### Grain Validation

Confirm that the physical table supports the intended analytical grain.

For example:

```text
Cohort
+
Wave
+
Race
+
Gender
```

should represent one analytical observation.

### Step 13 Deliverable

A validated Gold Delta table.

---

# 17. Step 14 — Begin Gold DML Specification

The Gold DML specification begins only after the Gold DDL has been reviewed and the physical table has been validated.

The DML specification defines:

* Source Silver tables
* RAND HRS variables
* Source-to-target mappings
* Lookups
* Transformations
* Calculations
* Aggregations
* Filtering
* Data-loading logic

### Handoff

```text
              GOLD DDL
                  │
                  ▼
          Create Gold Table
                  │
                  ▼
          Validate Structure
                  │
                  ▼
        APPROVED GOLD TABLE
                  │
                  ▼
              GOLD DML
                  │
                  ▼
        Define Source Data
                  │
                  ▼
          Define Transformations
                  │
                  ▼
            Load Gold Table
```

# 21. Key Architectural Principle

The HRS Gold development process follows a deliberate sequence:

> **Understand the analytical requirement → identify the required HRS data → define the physical Gold table → generate and validate the DDL → then develop the DML.**

The DDL process should not begin with SQL.

It should begin with an understanding of **what the analytical user needs the data to accomplish**.

The result is a controlled progression:

```text
Business Question
       │
       ▼
Analytical Requirements
       │
       ▼
HRS Variables
       │
       ▼
Gold Table Design
       │
       ▼
DDL Specification
       │
       ▼
DDL SQL
       │
       ▼
Gold Delta Table
       │
       ▼
Validated Structure
       │
       ▼
DML Specification
```

This separation allows the HRS Gold layer to be expanded consistently across analytical subject areas such as:

* Demographics
* Health
* Financial
* Employment
* Retirement
* Family
* Housing
* Social Security
* Other HRS analytical domains
