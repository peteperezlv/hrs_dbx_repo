# Silver Common Data Model (CDM)

The data model consist of 5 entities.
1. Cohort
2. Respondent
3. Wave
4. Demographics
5. Health

The Demographics and Health entities are used to classify the survey responses.

# The entity relationships are as follows.
1. One Cohort to many Repondent records
2. One Respondent to Many Demographics records
3. One Respondent to Many Health records
4. One Wave to many Demographic records
5. One Wave to many Health records

```text
                         ┌───────────────────┐
                         │      COHORT       │
                         └─────────┬─────────┘
                                 1 │
                                   │
                                   ▼ Many
                         ┌─────────┴─────────┐
                         │    RESPONDENT     │
                         └─────────┬─────────┘
                                   ▼
                     ┌─────────────┴────────────┐
                   1 │                        1 │
                     │                          │
                     ▼ Many                     ▼ Many
          ┌──────────┴────────┐       ┌─────────┴─────────┐
          │   DEMOGRAPHICS    │       │       HEALTH      │
          └──────────┬────────┘       └─────────┬─────────┘
                Many ▲                     Many ▲
                     │                          │
                     │ 1                        │ 1
          ┌──────────┴─────────┐       ┌────────┴─────────┐
          │        WAVE        │       │      WAVE        │
          └────────────────────┘       └──────────────────┘
```
