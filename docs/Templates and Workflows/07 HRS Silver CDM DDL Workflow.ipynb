# 2. Workflow Objective

The objective of this workflow is to provide a standardized process for designing, generating, creating, and validating HRS Silver CDM Delta tables before beginning the DML development process.

The workflow ensures that:

1. Business and data requirements are understood before SQL generation begins.
2. Source HRS variables are identified and documented.
3. Source-to-target mappings are defined.
4. The target table structure is formally specified.
5. The DDL is generated from the approved specification.
6. The generated DDL is reviewed and refined.
7. The Delta table is created in Databricks.
8. The resulting table structure is validated.
9. The target table is approved before DML development begins.

The DDL workflow establishes the physical target structure.

The subsequent DML workflow is responsible for transforming and loading RAND HRS data into that structure.

---

# 3. Workflow Overview - How do I create and validate the table?

The overall process is:

```text
                 HRS SILVER CDM
                  REQUIREMENTS
                       │
                       ▼
              Requirements Analysis
                       │
                       ▼
              DDL Specification
                       │
                       ▼
             Specification Review
                       │
                       ▼
             DDL SQL Generation
                       │
                       ▼
               DDL Code Review
                       │
                       ▼
             Execute DDL in Databricks
                       │
                       ▼
              Delta Table Created
                       │
                       ▼
              Table Structure Testing
                       │
                       ▼
              DDL Validation Complete
                       │
                       ▼
              DML Development Begins
```

---

# 4. Workflow Stages

The DDL workflow consists of six major stages:

1. Requirements Analysis
2. DDL Specification
3. DDL SQL Generation
4. DDL Code Review and Refinement
5. Delta Table Creation
6. Table Structure Validation

The DML workflow begins only after the DDL workflow has been successfully completed.

---

# 5. Step 1 — Requirements Analysis

The first step is to understand the business and data requirements for the HRS subject area.

## 5.1 Document the Use Case

Document the purpose of the Silver CDM subject-area table.

The use case should identify:

* Business purpose
* Subject area
* Intended analytical use
* Population represented
* Expected observation grain

Example:

```text
Subject Area: Demographics

Purpose:
Store respondent demographic observations from the RAND HRS
Longitudinal Dataset for analytical use.

Expected Grain:
One respondent per survey wave.
```

---

## 5.2 Identify Source HRS Variables

Identify the RAND HRS variables required to support the subject area.

The source variables should be documented using the RAND HRS Codebook.

For each variable, identify:

* Variable name
* Variable label
* RAND data type
* Wave applicability
* Whether the variable is wave-specific
* Whether the variable is wave-invariant
* Applicable source codes
* Missing-value conventions

---

## 5.3 Create the Source-to-Target Data Map

Create an initial source-to-target mapping.

The mapping should identify:

```text
Source Variable
      ↓
Target Column
      ↓
Target Data Type
      ↓
Transformation
```

Example:

```text
R1AGEY_E
    ↓
agey_e
    ↓
DECIMAL(10,2)
    ↓
Direct / CAST
```

The mapping created during requirements analysis becomes an input to the DDL and subsequent DML specifications.

---

## 5.4 Conduct Requirements Walkthrough

Review the requirements and source-to-target mapping with the appropriate stakeholders.

Confirm:

* Business purpose
* Target grain
* Required attributes
* Source variables
* Wave coverage
* Natural identifiers
* Parent-table relationships
* Data types
* Required audit fields

### Gate 1 — Requirements Approval

The workflow should not proceed to DDL specification development until the requirements have been reviewed and approved.

---

# 6. Step 2 — Create the DDL Specification

Once requirements are approved, create the subject-area DDL specification.

## 6.1 Review Architecture

Review:

`HRS DDL and DML Template Architecture`

The architecture document explains the relationship between:

* DDL specification
* DDL master template
* Generated DDL
* Target Delta table
* DML specification

---

## 6.2 Review the Master DDL Template

Review:

`HRS_DDL_Master_Template.ipynb`

The master template defines the standardized structure and requirements for generating DDL.

---

## 6.3 Create the Subject-Area DDL Specification

Create a subject-specific DDL specification.

Examples:

```text
HRS Demographics DDL Specification
HRS Health DDL Specification
HRS Employment DDL Specification
HRS Wealth DDL Specification
```

The subject-area specification must define:

* Target table
* Target catalog
* Target schema
* Table description
* Primary key
* Identity column
* Parent foreign keys
* Natural identifiers
* Business columns
* Audit columns
* Data types
* Nullability
* Comments
* Applicable constraints

---

## 6.4 Conduct DDL Specification Walkthrough

Review the completed DDL specification before generating SQL.

Verify:

* All required columns are present.
* Column names are correct.
* Data types are appropriate.
* Nullability is correct.
* Primary-key requirements are correct.
* Foreign-key relationships are correct.
* Natural identifiers are correctly identified.
* Audit columns are defined.
* Table and column descriptions are complete.

### Gate 2 — DDL Specification Approval

The DDL specification should be approved before SQL generation begins.

---

# 7. Step 3 — Generate DDL SQL

Once the DDL specification is approved, generate the SQL.

## 7.1 Provide the DDL Specification to the AI Assistant

Provide the approved subject-area DDL specification to the AI Assistant using:

`HRS_DDL_Master_Template.ipynb`

The AI Assistant generates the DDL based only on the approved specification.

---

## 7.2 Generate the Initial DDL

The generated SQL should contain the required DDL operations.

For example:

```text
DROP TABLE IF EXISTS
        ↓
CREATE TABLE
        ↓
Column Definitions
        ↓
Identity Column
        ↓
Constraints
        ↓
Comments
        ↓
USING DELTA
```

The generated DDL must not contain DML operations.

---

## 7.3 Review the Generated DDL

Review the generated SQL against the approved DDL specification.

Verify:

* Catalog
* Schema
* Table name
* Table type
* Storage format
* Columns
* Data types
* Nullability
* Identity column
* Primary key
* Foreign keys
* Comments
* Constraints

---

# 8. Step 4 — DDL Code Review and Refinement

The generated DDL should be reviewed before execution.

## 8.1 Code Review

Compare:

```text
DDL Specification
        │
        ▼
Generated DDL
```

Every structural requirement in the specification should be represented correctly in the generated SQL.

---

## 8.2 Resolve Issues

If issues are identified:

```text
Generated DDL
      │
      ▼
Issue Identified
      │
      ▼
Review with AI Assistant
      │
      ▼
Refine DDL
      │
      ▼
Review Again
```

The AI Assistant must not introduce requirements that are not contained in the approved specification.

---

## 8.3 Final DDL Approval

Once the generated DDL matches the approved specification, designate the SQL as the candidate deployment version.

### Gate 3 — DDL Code Approval

The DDL must pass code review before execution in Databricks.

---

# 9. Step 5 — Create the Delta Table

Execute the approved DDL in the Databricks environment.

## 9.1 Execute DDL

Run the generated SQL using the designated Databricks environment.

The expected result is:

```text
DDL SQL
   │
   ▼
Databricks
   │
   ▼
Managed Delta Table
```

---

## 9.2 Confirm Successful Creation

Confirm that the table was successfully created.

Verify:

* No SQL errors
* Table exists
* Correct catalog
* Correct schema
* Correct table name

---

## 9.3 Confirm Delta Table Properties

Confirm that the target is:

```text
Format: Delta
Type: Managed Table
```

Confirm that the table is located in the expected Databricks-managed storage location.

The physical location should be documented for verification purposes but should not be hard-coded into the DDL for a managed table.

---

# 10. Step 6 — Validate the Delta Table Structure

After the table has been created, validate the physical table against the approved DDL specification.

## 10.1 Structural Validation

Validate:

* Table exists
* Correct catalog
* Correct schema
* Correct table name
* Delta format
* Managed table

---

## 10.2 Column Validation

Validate:

* Required columns exist
* Column names are correct
* Data types are correct
* Nullability is correct
* Identity column exists
* Identity column has the correct definition

---

## 10.3 Constraint Validation

Validate:

* Primary key exists
* Respondent foreign key exists
* Wave foreign key exists
* Applicable constraints exist

---

## 10.4 Comment Validation

Validate:

* Table comment exists
* Required column comments exist

---

## 10.5 Parent Table Validation

Confirm that the required parent tables exist:

```text
dev_catalog.slv_cdm_hrs.hub_respondent

dev_catalog.slv_cdm_hrs.dim_wave
```

Confirm that their required parent keys exist:

```text
hub_respondent.respondent_id

dim_wave.wave_id
```

---

# 11. DDL Unit Testing

The target table should be tested before beginning DML development.

## 11.1 Test Identity Column

Confirm that the identity column exists and is configured as a system-generated identity column.

The DML must not be responsible for generating this value.

---

## 11.2 Test Foreign-Key Structure

Confirm that:

```text
fact_<subject>.respondent_id
        ↓
hub_respondent.respondent_id
```

and:

```text
fact_<subject>.wave_id
        ↓
dim_wave.wave_id
```

are defined as specified.

---

## 11.3 Test Column Data Types

Compare every target column against the DDL specification.

Example:

```text
Specification       Actual Table
─────────────────────────────────
agey_e DECIMAL      agey_e DECIMAL
raracem TINYINT     raracem TINYINT
wave_number STRING  wave_number STRING
```

Any discrepancy must be resolved before proceeding.

---

## 11.4 Test Nullability

Confirm that required columns are defined as `NOT NULL` and optional business attributes are nullable according to the specification.

---

## 11.5 Test Table Comments

Confirm that the table and required columns contain the specified comments.

---

# 13. DDL-to-DML Handoff

Once the DDL validation gate has passed, the target table is considered ready for DML development.

The handoff is:

```text
              DDL WORKFLOW
                   │
                   ▼
          Requirements Approved
                   │
                   ▼
          DDL Specification
                   │
                   ▼
             DDL Generated
                   │
                   ▼
            DDL Reviewed
                   │
                   ▼
          Delta Table Created
                   │
                   ▼
          Structure Validated
                   │
                   ▼
          DDL APPROVED
                   │
                   ▼
          DML WORKFLOW BEGINS
```

The DML workflow should not begin until the DDL workflow has successfully completed its validation gate.

---

# 14. DML Workflow Starting Point

The DML workflow begins with an existing and validated Silver CDM target table.

The DML process will then define:

```text
Source RAND HRS Data
        │
        ▼
Natural Identifier Resolution
        │
        ├── HHIDPN → respondent_id
        │
        └── wave_number → wave_id
        │
        ▼
Wide-to-Long Transformation
        │
        ▼
Source-to-Target Mapping
        │
        ▼
Data-Type Conversion
        │
        ▼
Business Rules
        │
        ▼
NULL / Missing-Value Rules
        │
        ▼
Audit Columns
        │
        ▼
Business-Grain Validation
        │
        ▼
INSERT INTO Target Table
```

The DML workflow is documented separately in:

`HRS Silver CDM Master DML Functional Specification`

# 16. Key Workflow Principles

### Principle 1 — Requirements Before Code

Do not generate SQL until the business and data requirements have been documented.

### Principle 2 — Specification Before Generation

Do not generate final DDL until the subject-area DDL specification has been reviewed.

### Principle 3 — Review Before Deployment

Do not execute generated DDL until it has been reviewed against the specification.

### Principle 4 — Validate Before DML

Do not begin DML development until the target Delta table has been successfully created and validated.

### Principle 5 — Specification Is the Authority

The approved specification is the authoritative definition of the required table structure.

### Principle 6 — No Invented Requirements

The AI Assistant must not invent columns, keys, constraints, transformations, or business rules that are not defined in the applicable specification.

# 17. Final Workflow

The complete DDL workflow is:

```text
1. REQUIREMENTS ANALYSIS
          │
          ▼
   Document Use Case
          │
          ▼
   Identify HRS Variables
          │
          ▼
   Create Source-to-Target Map
          │
          ▼
   Requirements Approval
          │
          ▼
2. DDL SPECIFICATION
          │
          ▼
   Review Architecture
          │
          ▼
   Review DDL Master Template
          │
          ▼
   Create Subject DDL Specification
          │
          ▼
   DDL Specification Approval
          │
          ▼
3. DDL GENERATION
          │
          ▼
   AI Generates DDL
          │
          ▼
   Review Generated SQL
          │
          ▼
4. DDL CODE REVIEW
          │
          ▼
   Refine / Correct SQL
          │
          ▼
   DDL Code Approval
          │
          ▼
5. DELTA TABLE CREATION
          │
          ▼
   Execute DDL in Databricks
          │
          ▼
   Confirm Table Creation
          │
          ▼
6. TABLE VALIDATION
          │
          ▼
   Validate Structure
          │
          ▼
   Validate Columns
          │
          ▼
   Validate Keys
          │
          ▼
   Validate Data Types
          │
          ▼
   Run DDL Unit Tests
          │
          ▼
       DDL APPROVED
          │
          ▼
     DML WORKFLOW
```

---

# 18. DDL Workflow Completion

The DDL workflow is considered complete when:

1. The requirements have been approved.
2. The subject-area DDL specification has been approved.
3. The DDL SQL has been generated.
4. The DDL SQL has been reviewed and approved.
5. The Delta table has been successfully created.
6. The table structure matches the approved DDL specification.
7. Required keys and constraints are present.
8. Required data types and nullability are correct.
9. DDL unit tests have passed.

At this point, the target Silver CDM table is ready for the **DML transformation and loading workflow**.
