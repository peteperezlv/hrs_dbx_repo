
# High Level SQL Generation Flow

1. Inputs: The user asks the AI Agent to do some work.  The uer provides three files to the AI Agent.
    * The prompt file which provides the processing instrutions.
    * The Use Case document which provides the list of business questions.
    * The RAND HRS Codebook.  This the source data definitions and dictionary.

2. Processing: The AI Agent then that the Inputs to process the request.

3. Outputs: The AI Agent generates the SQL outputs.

4. Data Analyst manually runs the generated DDL and validates the delta tables.

# The DDL Workflow
```text


                                            ┌─────DDL Prompt File (Instructions)
                                            ┌─────Use Case (Requirements)
                                            ┌─────HRS Code Book (Data Dictionary)
                                            ▼
                                            | 
                                            | Inputs
    ┌────────────────────┐      ┌───────────┴───────────┐     
    │    Data Analyst    |─────>|          AI           | Processing
    |       Request      │      |         AGENT         │
    └──────────┬─────────┘      └───────────┬───────────┘
               |                            |
               |                            ▼ Outputs
               |                            | 
               |                 ┌──────────┴───────────     
               |                 |  SQL DDL Code
               |                 └──────────┬───────────
               |                            |
               |                            ▼ 
               |                            | 
               |                 ┌──────────┴───────────┐     
               └────────────────>|  Manually Execute    |
                                 |        DDL           │
                                 └──────────────────────┘



```
## DDL Workflow Steps
1. Manual - Create the Master Use Case with all the business questions.
    * Use this notebook: `\hrs_dbx_repo\docs\Use Cases\UC-DA-MASTER.ipynb`
2. Manual - Identify the data source.  For this project the source data is the RAND HRS Longitudinal Dataset.
3. Manual - Ingest and transform the data source to a delta table.  Here we simply convert the the SPSS file to a Delta Table.
    * Use the prepared notebook for this step: `\hrs_dbx_repo\notebooks\ingestion\convert_rand_logitudinal_spss_file.ipynb`
4. Manual - Create the Conceptual Entity Relationship Model. 
    * You can reference the data in notebook: `\hrs_dbx_repo\docs\Data Model\Silver_CDM_Data_Model.ipynb`    
5. Manual - Create the DDL Generation Prompt to support the Master Use Case.
    * The Use Case can be found here: `\hrs_dbx_repo\docs\Use Cases\UC-DA-MASTER.ipynb`
6. AI - Run the DDL SQL Generation Prompt.
    * The prompt is documented here: `\hrs_dbx_repo\docs\Prompts\00 Prompt Generate Silver DDL Master UC.ipynb`
7. Manual - Execute the DDL SQL Scripts
8. Manual - Test and refine the Use Case and Prompts.


# The DML Workflow
```text


                                            ┌─────DML Prompt File (Instructions)
                                            ┌─────Use Case (Requirements)
                                            ┌─────HRS Code Book (Data Dictionary)
                                            ┌─────HRS Delta Table (ingested and converted from HRS Source)
                                            ┌─────HRS Silver Delta Tables
                                            ▼
                                            | 
                                            | Inputs
    ┌────────────────────┐      ┌───────────┴───────────┐     
    │    Data Analyst    |─────>|          AI           | Processing
    |       Request      │      |         AGENT         │
    └──────────┬─────────┘      └───────────┬───────────┘
               |                            |
               |                            ▼ Outputs
               |                            | 
               |                 ┌──────────┴───────────     
               |                 |  SQL DML Code
               |                 └──────────┬───────────
               |                            |
               |                            ▼ 
               |                 ┌──────────┴───────────┐     
               └────────────────>|  Manually Execute    |
                                 |        DML           │
                                 └──────────────────────┘


```
