
# High Level SQL Generation Flow

1. Inputs: The user asks the AI Agent to do some work.  The uer provides three files to the AI Agent.
    * The prompt file which provides the processing instrutions.
    * The Use Case document which provides the list of business questions.
    * The RAND HRS Codebook.  This the source data definitions and dictionary.

2. Processing: The AI Agent then that the Inputs to process the request.

3. Outputs: The AI Agent generates the SQL outputs.

4. Data Analyst manually runs the generated DDL and validates the delta tables.

# The DDL Workflow
```text


                                            ┌─────DDL Prompt File (Instructions)
                                            ┌─────Use Case (Requirements)
                                            ┌─────HRS Code Book (Data Dictionary)
                                            ▼
                                            | 
                                            | Inputs
    ┌────────────────────┐      ┌───────────┴───────────┐     
    │    Data Analyst    |─────>|          AI           | Processing
    |       Request      │      |         AGENT         │
    └──────────┬─────────┘      └───────────┬───────────┘
               |                            |
               |                            ▼ Outputs
               |                            | 
               |                 ┌──────────┴───────────     
               |                 |  SQL DDL Code
               |                 └──────────┬───────────
               |                            |
               |                            ▼ 
               |                            | 
               |                 ┌──────────┴───────────┐     
               └────────────────>|  Manually Execute    |
                                 |        DDL           │
                                 └──────────────────────┘



```
## DDL Workflow Steps
1. User - Create the Master Use Case with all the business questions.
    * Use this notebook: `\hrs_dbx_repo\docs\Use Cases\UC-DA-MASTER.ipynb`
2. User - Identify the data source.  For this project the source data is the RAND HRS Longitudinal Dataset.
3. User - Ingest and transform the data source to a delta table.  Here we simply convert the SPSS file to a Delta Table.
    * Use the prepared notebook for this step: `\hrs_dbx_repo\notebooks\ingestion\convert_rand_logitudinal_spss_file.ipynb`
4. User - Create the Conceptual Entity Relationship Model. 
    * You can reference the data in notebook: `\hrs_dbx_repo\docs\Data Model\Silver_CDM_Data_Model.ipynb`    
5. User - Create the AI Agent Prompt to support the Master Use Case.
    * The Prompt file can be found here: `\hrs_dbx_repo\docs\Prompts\00 Prompt Generate Silver DDL Master UC.ipynb`
6. AI - Direct the Ai Agent to use the DDL SQL Generation Prompt.
    * Directions given to the Ai Agent
        - You are a senior Databricks Data Analyst.  
        - Your task is to generate the Silver layer DDL scripts to support data analytics and visualizations. 
        - Use this prompt file for detailed directions: `\hrs_dbx_repo\docs\Prompts\00 Prompt Generate Silver DDL Master`

7. User - Execute the DDL SQL Scripts
8. User - Test and refine the Use Case and Prompts.
