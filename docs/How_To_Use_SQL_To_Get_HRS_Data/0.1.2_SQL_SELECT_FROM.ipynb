## Purpose

The purpose of this guide is to introduce you to the two most fundamental building blocks of SQL: the SELECT and FROM clauses. ## Objective

The objective of this section is to teach you how to write simple two-line instructions to look inside our relational database. By working through these introductory examples, you will learn how to point your query to a specific table using FROM and choose the exact columns you want to display on your screen using SELECT, giving you complete control over what information you pull into view.

## The Fundamental Rule: What and Where
Every single search you perform in our database starts with these two core instructions. You can think of them as a basic fill-in-the-blank sentence:

`SELECT` [What data do you want to see?]

`FROM`   [Where is the table or list located?];



%md
## List all the HRS COHORTS records.
### The HRS_COHORT table is a dimension table in our data model.  It hold the descriptive data for each COHORT.

SELECT statement Breakdown
1. SELECT (What you want to see?)
    * use the **"*"** identifier to tell the database to return all records in the table.
2. FROM (Where is the data stored?)
    * you need to specify the database path (**staging_catalog.slv_cdm_hrs**) and the table (**hrs_cohort**)


```python
-- HRS_ COHORT dimension table
SELECT * 
FROM staging_catalog.slv_cdm_hrs.hrs_cohort
```

```python
-- HRS_WAVE dimension table
SELECT * 
FROM staging_catalog.slv_cdm_hrs.hrs_wave
```

## Count number of records in the RESPONDENT table
SELECT statement Breakdown
1. SELECT (What you want to see?)
    * use the count(*) aggregate function to tell the database to give you a total count of recrods.
2. FROM (Where is the data stored?)
    * you need to specify the database path (**staging_catalog.slv_cdm_hrs**) and the table (**hrs_respondent**)

Note: Each respondent is uiquely identified by the HHIDPN number.


```python
%sql
-- Get a count of all the records in the hrs_resondent table.
SELECT count(*) 
FROM staging_catalog.slv_cdm_hrs.hrs_respondent
```

```python
%sql
-- Get a count of all the unique HHIDPN valuess hrs_resondent table.
-- The cound should be that same as the number of records in the table.
-- This is because the HHIDPN is a unique identifier for each household.

SELECT count(hhidpn) 
FROM staging_catalog.slv_cdm_hrs.hrs_respondent
```

```python
%sql
-- Select 10 records from the hrs_respondent table.
-- Here we just add the LIMIT clause to the end of the query.  
-- We have aked SQL to return only 10 records
-- You will see that the HHIDPN is unique for each record and points to the HACOHORT.
SELECT *
FROM staging_catalog.slv_cdm_hrs.hrs_respondent
LIMIT 10
```
