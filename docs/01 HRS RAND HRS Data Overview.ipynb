# 2. Introduction

The **Health and Retirement Study (HRS)** is one of the most important longitudinal studies of aging and retirement in the United States.

The study follows Americans as they move through later working life, retirement, and older age. It collects information across multiple dimensions of an individual's life, including:

* Demographics
* Physical health
* Mental and cognitive health
* Health care utilization
* Health care expenditures
* Employment
* Retirement
* Income
* Wealth
* Pensions
* Health insurance
* Family relationships
* Disability and functional limitations
* Expectations and retirement planning
* Psychosocial characteristics

The HRS began in 1992 and interviews participants approximately every two years. The study now represents a large longitudinal research resource containing information on more than 20,000 Americans.

The RAND HRS Longitudinal File is a cleaned, processed, and streamlined data product derived from the HRS and designed to make a substantial portion of the HRS data easier for researchers to use.

# 3. Who Conducts the HRS?

## 3.1 University of Michigan

The Health and Retirement Study is conducted by the **Institute for Social Research (ISR) at the University of Michigan**.

The HRS is supported primarily by the **National Institute on Aging (NIA)**, with additional support from the **Social Security Administration (SSA)**.

The University of Michigan is responsible for the underlying HRS study, including:

* Survey design
* Data collection
* Questionnaire development
* Sample management
* Data processing
* Data documentation
* Public data distribution
* Restricted-data programs
* Longitudinal tracking

The HRS is therefore the underlying research study.

# 4. What Is the RAND HRS?

The **RAND HRS** is a family of data products created by the **RAND Center for the Study of Aging** using information from the Health and Retirement Study.

RAND works with HRS data to produce cleaned and standardized files that make many HRS variables easier for researchers to use.

The RAND HRS Longitudinal File is particularly useful because it provides a consistent longitudinal representation of many HRS variables across survey waves.

The relationship can be summarized as:

```text
                    UNIVERSITY OF MICHIGAN
                           │
                           ▼
                   HEALTH AND RETIREMENT
                       STUDY (HRS)
                           │
                           │
                           ▼
                     HRS Survey Data
                           │
                           ▼
                  RAND Center for the
                     Study of Aging
                           │
                           ▼
                        RAND HRS
                     Data Products
                           │
                           ▼
                RAND HRS Longitudinal
                       File
```

Therefore:

> **HRS is the underlying longitudinal study; RAND HRS is a processed data product derived from HRS data.**

# 5. Purpose of the HRS

The original purpose of the HRS was to provide longitudinal information that would allow researchers to study the transition from working life into retirement and older age.

The study has expanded substantially since 1992.

Today, HRS supports research into the interaction among:

* Health
* Employment
* Retirement
* Income
* Wealth
* Family
* Health care
* Disability
* Cognition
* Social circumstances

The National Institute on Aging describes HRS as a resource for understanding changes in labor-force participation and health transitions as individuals approach and move through retirement and later life.

---

# 6. Why the HRS Is Different From a Typical Health Survey

One of the most valuable characteristics of HRS is that it is **longitudinal**.

A traditional cross-sectional survey may provide a snapshot:

```text
Person → Health Status → 2026
```

A longitudinal study can provide:

```text
Person
  │
  ├── Wave 1 → Health / Work / Income
  │
  ├── Wave 2 → Health / Work / Income
  │
  ├── Wave 3 → Health / Work / Income
  │
  ├── Wave 4 → Health / Work / Income
  │
  └── ...
```

This allows researchers to study **change over time**.

For example:

* Did health decline before retirement?
* Did retirement affect health?
* How does income change after retirement?
* How does disability affect employment?
* How does wealth affect access to health care?
* How do health conditions evolve as individuals age?

These are fundamentally different questions from simply asking how many people have a condition at a particular point in time.

# 7. How HRS Data Are Collected

The HRS uses a **multi-stage area probability sample** designed to produce a representative sample of the U.S. population in the study's target age range.

The survey has historically used a combination of **in-person and telephone interviewing**.

The survey is conducted approximately every two years.

The HRS questionnaire has evolved over time, and the data collection technology has also changed. For example, the survey moved from Surveycraft to Blaise beginning in 2002.

# 9. HRS Cohorts

An important concept when working with HRS data is the **cohort**.

A cohort is a group of individuals defined primarily by their birth years and the period in which they entered the study.

The RAND HRS Longitudinal File currently describes eight cohorts:

| Cohort             | Approximate Birth Years |            Entry |
| ------------------ | ----------------------: | ---------------: |
| HRS                |               1931–1941 |             1992 |
| AHEAD              |         1923 or earlier |             1993 |
| CODA               |               1924–1930 |             1998 |
| War Baby           |               1942–1947 |             1998 |
| Early Baby Boomer  |               1948–1953 |             2004 |
| Mid Baby Boomer    |               1954–1959 |             2010 |
| Late Baby Boomer   |               1960–1965 |             2016 |
| Early Generation X |               1966–1971 | Newer HRS cohort |

The original HRS cohort consisted of individuals born from 1931 through 1941 who were approximately 51–61 years old when first interviewed in 1992.

The AHEAD study was added in 1993 and focused on individuals born before 1924 who were age 70 or older at the time.

The continuing HRS cohorts have subsequently been expanded to represent newer generations as they approach later working life and older age. The current HRS data also include an Early Generation X cohort in the newer releases.

---

# 10. Cohort Versus Wave

These two concepts are frequently confused.

## Cohort

A cohort identifies a group of people based primarily on their birth years and study entry.

```text
COHORT
   │
   ├── HRS
   ├── AHEAD
   ├── CODA
   ├── War Baby
   ├── Early Baby Boomer
   ├── Mid Baby Boomer
   ├── Late Baby Boomer
   └── Early Generation X
```

## Wave

A wave identifies a particular period of HRS data collection.

```text
WAVE 1 → 1992
WAVE 2 → 1993
WAVE 3 → 1994
WAVE 4 → 1995
WAVE 5 → 1996
...
```

Beginning with the later period of the study, core interviews generally occur biennially.

The HRS documentation provides questionnaires and data products organized across these waves.

---

# 11. Respondents

A **respondent** is an individual who has participated in the HRS.

The RAND HRS Longitudinal File includes individuals who were interviewed by HRS at least once.

This includes:

* Age-eligible respondents
* Spouses of age-eligible respondents
* Individuals who became spouses between survey waves
* Other individuals included through the study's longitudinal sample design

The RAND documentation specifically notes that the longitudinal file includes anyone interviewed by HRS at least once.

---

# 12. Respondent Identification

The HRS uses identifiers that allow researchers to follow individuals across waves.

For the RAND HRS Longitudinal File used in this project, the respondent-level natural identifier is:

```text
HHIDPN
```

Conceptually:

```text
HHIDPN
   │
   ▼
Respondent
   │
   ├── Wave 1
   ├── Wave 2
   ├── Wave 3
   └── ...
```

The HRS also provides tracker products specifically designed to facilitate the use of HRS data within and across waves.

---

# 13. Household and Respondent Concepts

HRS contains both household-related and individual-level information.

This distinction is important when designing analytical data models.

A household may contain:

```text
Household
   │
   ├── Respondent
   │
   └── Spouse / Partner
```

Some information is collected at the household level, while other information is specific to an individual respondent.

The RAND HRS Longitudinal File uses a respondent-oriented structure that brings together relevant household and spousal information for analysis.

# 14. The RAND HRS Longitudinal File

The RAND HRS Longitudinal File is particularly important for this project.

RAND describes it as a:

> cleaned, processed, and streamlined collection of variables derived from HRS.

It primarily processes information from HRS Core and Exit Interviews and incorporates information from additional HRS files such as Tracker, Region and Mobility, and Master ID files.

The current RAND product listing includes a **RAND HRS Longitudinal File 2022**, with a May 2025 release listed by HRS.

# 16. The HRS Data Model Used in This Project

The source RAND HRS data is fundamentally wide and longitudinal.

For example:

```text
HHIDPN
R1AGEY_E
R2AGEY_E
R3AGEY_E
...
R16AGEY_E
```

The Silver CDM reorganizes this structure into a more relational respondent-by-wave design.

Conceptually:

```text
                    RESPONDENT
                         │
                         │
                         ▼
                   respondent_id
                         │
             ┌───────────┼───────────┐
             │           │           │
             ▼           ▼           ▼
           Wave 1      Wave 2      Wave 3
             │           │           │
             ▼           ▼           ▼
        Demographics Demographics Demographics
        Health       Health       Health
        Income       Income       Income
        Employment   Employment   Employment
```

This design makes longitudinal SQL analysis substantially easier.

The HRS is particularly valuable for public health because health does not exist independently of the social and economic environment.

For example:

```text
Health
  │
  ├── Income
  ├── Employment
  ├── Retirement
  ├── Wealth
  ├── Insurance
  ├── Family
  ├── Disability
  └── Health Care
```

The longitudinal structure allows researchers to examine how these factors interact over time.

This creates opportunities to move beyond simple prevalence statistics toward analysis of:

* Risk factors
* Health trajectories
* Health disparities
* Health care utilization
* Functional decline
* Economic vulnerability
* Retirement transitions
* Long-term consequences

# 26. Important Considerations When Using HRS

HRS is a powerful research resource, but analysts must understand the study design before drawing conclusions.

Important considerations include:

* Survey weights
* Complex sample design
* Cohort differences
* Wave differences
* Respondent attrition
* Missing data
* Imputed values
* Variable definitions
* Changes in questionnaires
* Changes in data collection methods
* Cross-wave comparability

HRS provides documentation and user guides to help researchers understand these issues.

---

# 27. Survey Weights

HRS is designed to support population-representative analyses when the appropriate weighting methodology is used.

The study uses sampling weights, including special considerations associated with its sample design and population representation.

The HRS oversamples certain populations, including Black and Hispanic respondents and residents of Florida, and provides weighting variables to support appropriate population estimates.

Therefore:

> **Analysts should not automatically interpret an unweighted HRS sample statistic as a national population estimate.**

The appropriate survey weights and analytical methods should be determined based on the research question and HRS documentation.

---

# 28. Public Versus Restricted Data

Not every HRS data product is publicly available.

HRS provides public data products to registered users while certain sensitive data require additional access procedures.

The RAND HRS Longitudinal File used as the foundation for this project does not include restricted HRS data.

Researchers should always verify the access restrictions associated with the specific HRS data product they intend to use.

# 29. HRS Data and the Silver CDM Project

The purpose of the HRS Silver CDM project is not to replace the official HRS or RAND HRS data products.

Instead, it creates an analytical data model optimized for:

* Data engineering
* SQL analysis
* Longitudinal analysis
* Data integration
* Visualization
* Reusable analytical workflows

The relationship is:

```text
Official HRS Study
       │
       ▼
RAND HRS Data Product
       │
       ▼
HRS Bronze Delta
       │
       ▼
HRS Silver CDM
       │
       ▼
Analytics
       │
       ▼
Policy Research
```

The official HRS and RAND documentation remain the authoritative sources for understanding the underlying survey and its variables.

# 33. Summary

The Health and Retirement Study provides an unusually rich longitudinal view of aging in the United States.

Its strength comes from the combination of:

```text
Large Population
       +
Longitudinal Follow-up
       +
Multiple Birth Cohorts
       +
Health Information
       +
Economic Information
       +
Employment Information
       +
Family Information
       +
Health Care Information
       +
Cognitive / Psychosocial Information
       =
Multidisciplinary Aging Research Resource
```

The University of Michigan conducts and maintains the HRS, while RAND's Center for the Study of Aging produces user-friendly RAND HRS data products from HRS information.

For this project, the **RAND HRS Longitudinal File** provides the starting point for building a structured Silver CDM.

The resulting CDM can then provide a consistent foundation for:

* Public health analysis
* Hospital and health-system research
* State agency analysis
* Federal policy research
* Aging research
* Health disparities research
* Longitudinal health analysis
* Economic and retirement research
* Data visualization

The ultimate objective is to transform a complex longitudinal research dataset into an accessible analytical environment where researchers can move from **HRS survey data → structured data → analysis → evidence → policy insight**.
