# 2. Introduction

The **Health and Retirement Study (HRS)** is one of the most important longitudinal studies of aging and retirement in the United States.

The study follows individuals and their families as they move through later working life, retirement, and older age. It collects information across multiple dimensions of a person's life, including:

* Demographics
* Physical health
* Cognitive health
* Functional limitations
* Health care
* Employment
* Retirement
* Income
* Wealth
* Pensions
* Social Security
* Health insurance
* Family structure
* Retirement expectations
* Psychosocial characteristics

The HRS began in 1992 and has continued to expand its coverage of the U.S. population.

The **RAND HRS Longitudinal File** is a major data product derived from the HRS and is designed to provide researchers with a more accessible, standardized representation of HRS information.

The RAND HRS presentation describes the Longitudinal File as an **easy-to-use entry point to HRS data** and notes that many users use it as the backbone of their analyses before incorporating additional HRS or RAND data products as needed.

---

# 3. The University of Michigan and the HRS

The Health and Retirement Study is conducted by the **University of Michigan's Institute for Social Research**.

The study is primarily supported by the **National Institute on Aging (NIA)**, with additional support from the **Social Security Administration (SSA)**.

The University of Michigan is responsible for the underlying HRS study, including:

* Survey design
* Questionnaire development
* Sample management
* Data collection
* Longitudinal tracking
* Data processing
* Documentation
* Public data distribution

The HRS is therefore the **underlying research study**, while RAND HRS products are processed research datasets derived from HRS information.

# 4. What Is RAND HRS?

The **RAND HRS** refers to a collection of data products developed by the RAND Center for the Study of Aging using HRS data.

RAND processes HRS information to create standardized and analytically useful datasets.

One of the most important of these products is the:

> **RAND HRS Longitudinal File**

The RAND HRS Longitudinal File combines information from multiple HRS waves into a single longitudinal dataset.

It contains derived variables based on information from the HRS Core data, Tracker data, and other HRS sources.

The relationship can be represented as:

# 6. Why the RAND HRS Longitudinal File Is Important

The RAND HRS Longitudinal File is particularly useful as a starting point because it provides a consistent longitudinal framework.

The presentation identifies several important characteristics:

* It combines information across HRS waves.
* It provides derived variables.
* It contains information across multiple subject areas.
* It can be merged with other RAND HRS products.
* It can be combined with public HRS Core data modules when additional variables are required.

This makes the Longitudinal File particularly well suited to the architecture being developed in this project.

The project therefore uses the RAND HRS Longitudinal File as the primary source for the initial HRS Silver CDM.

## 6.1 Why the HRS Is Different From a Typical Health Survey

One of the most valuable characteristics of HRS is that it is **longitudinal**.

A traditional cross-sectional survey may provide a snapshot:

```text
Person → Health Status → 2026
```

A longitudinal study can provide:

```text
Person
  │
  ├── Wave 1 → Health / Work / Income
  │
  ├── Wave 2 → Health / Work / Income
  │
  ├── Wave 3 → Health / Work / Income
  │
  ├── Wave 4 → Health / Work / Income
  │
  └── ...
```
## 6.2 How HRS Data Are Collected

The HRS uses a **multi-stage area probability sample** designed to produce a representative sample of the U.S. population in the study's target age range.

The survey has historically used a combination of **in-person and telephone interviewing**.

The survey is conducted approximately every two years.

The HRS questionnaire has evolved over time, and the data collection technology has also changed. For example, the survey moved from Surveycraft to Blaise beginning in 2002.

# 7. What Is a HRS Wave?

A **wave** represents a period of HRS data collection.

However, there is an important distinction between an HRS survey year and the RAND HRS wave numbering system.

The RAND HRS presentation specifically notes that some RAND waves correspond to multiple HRS survey years.

For example:

| RAND Wave | HRS Survey Years | Cohorts                                   |
| --------- | ---------------- | ----------------------------------------- |
| 1         | 1992             | HRS, AHEAD                                |
| 2         | 1993, 1994       | HRS, AHEAD                                |
| 3         | 1995, 1996       | HRS, AHEAD                                |
| 4         | 1998             | HRS, AHEAD, CODA, War Baby                |
| 5         | 2000             | Multiple cohorts                          |
| 6         | 2002             | Multiple cohorts                          |
| 7         | 2004             | HRS, AHEAD, CODA, War Baby, EBB           |
| 8         | 2006             | Multiple cohorts                          |
| 9         | 2008             | Multiple cohorts                          |
| 10        | 2010             | HRS, AHEAD, CODA, War Baby, EBB, MBB      |
| 11        | 2012             | Multiple cohorts                          |
| 12        | 2014             | Multiple cohorts                          |
| 13        | 2016             | HRS, AHEAD, CODA, War Baby, EBB, MBB, LBB |
| 14        | 2018             | Multiple cohorts                          |
| 15        | 2020             | Multiple cohorts                          |

The presentation emphasizes that the RAND convention calls the AHEAD 1993 survey **Wave 2A** and the HRS 1994 survey **Wave 2H**. Similarly, AHEAD 1995 and HRS 1996 are referred to as **Wave 3A** and **Wave 3H**.

### Important

For this project, **wave_number should be treated as a natural identifier rather than simply assuming that a wave number equals a calendar year**.

The Silver CDM therefore uses:

```text
wave_number → wave_id
```

where:

* `wave_number` is the natural identifier derived from the source data.
* `wave_id` is the system-generated surrogate key in the Silver CDM.

# 8. HRS Cohorts

A **cohort** is a group of individuals associated with a particular birth-year range and study entry period.

The HRS has added cohorts over time to maintain representation of individuals entering the target age range.

Major cohorts include:

| Cohort   | General Description                             |
| -------- | ----------------------------------------------- |
| HRS      | Original Health and Retirement Study cohort     |
| AHEAD    | Older Americans included beginning in 1993      |
| CODA     | Children of the Depression                      |
| War Baby | Individuals born during the World War II period |
| EBB      | Early Baby Boomers                              |
| MBB      | Mid Baby Boomers                                |
| LBB      | Late Baby Boomers                               |
| Gen X    | Newer generations entering the study            |

The RAND presentation notes that new cohorts are continually added to fill age gaps and maintain representation of populations reaching approximately age 51.

---

# 9. Cohort Versus Wave

These concepts should not be confused.

### Cohort

A cohort identifies **who entered the study based primarily on birth year and study design**.

### Wave

A wave identifies **when the respondent was observed/interviewed within the HRS longitudinal study**.

Conceptually:

```text
COHORT
   │
   ├── HRS
   ├── AHEAD
   ├── CODA
   ├── War Baby
   ├── Early Baby Boomer
   ├── Mid Baby Boomer
   └── Late Baby Boomer
            │
            ▼
         RESPONDENT
            │
     ┌──────┼──────┐
     ▼      ▼      ▼
  Wave 1  Wave 2  Wave 3
     │      │      │
     ▼      ▼      ▼
   Data   Data   Data
```

This distinction becomes important when designing longitudinal analytical tables.

---

# 10. Respondents

A **respondent** is an individual participating in the HRS.

The longitudinal nature of HRS means that the same respondent may appear in multiple waves.

For example:

```text
Respondent
   │
   ├── Wave 1
   ├── Wave 2
   ├── Wave 3
   ├── Wave 4
   └── ...
```

The RAND HRS Longitudinal File includes individuals who were interviewed by HRS at least once.

---

# 11. Respondent Identification

A key identifier in the RAND HRS Longitudinal File is:

```text
HHIDPN
```

HHIDPN provides a unique respondent-level identifier that can be used to merge the RAND HRS Longitudinal File with other HRS and RAND datasets.

For the Silver CDM architecture:

```text
HHIDPN
   │
   ▼
hub_respondent
   │
   ▼
respondent_id
```

The design intentionally separates the **natural source identifier** from the **system-generated warehouse key**.

Therefore:

* `HHIDPN` = natural identifier
* `respondent_id` = system-generated surrogate key

This allows the Silver CDM to maintain stable relational relationships while preserving the original HRS identifier for source-system traceability.

---

# 12. Respondent-by-Wave Structure

The RAND HRS source is organized primarily as a wide longitudinal dataset.

For example, a respondent's age may be represented as:

```text
HHIDPN
R1AGEY_E
R2AGEY_E
R3AGEY_E
...
R16AGEY_E
```

The Silver CDM transforms this structure into a respondent-by-wave model:

```text
HHIDPN
   │
   ▼
respondent_id
   │
   ├──────────────┐
   │              │
   ▼              ▼
Wave 1          Wave 2
   │              │
   ▼              ▼
Demographics   Demographics
Health         Health
Income         Income
...
```

This transformation makes SQL analysis easier because observations are represented vertically rather than requiring analysts to repeatedly work with wave-specific columns.

# 13. RAND HRS Variable Naming Conventions

One of the most important concepts for working with the RAND HRS Longitudinal File is its variable naming convention.

Many variables contain a wave identifier.

For example:

```text
R1AGEY_E
R2AGEY_E
R3AGEY_E
```

The `R1`, `R2`, and `R3` components identify the corresponding wave.

However, **not every variable is wave-specific**.

The RAND documentation identifies several important naming patterns.

### Wave-Specific Variables

Example:

```text
R1AGEY_E
R2AGEY_E
R3AGEY_E
```

### Wave-Invariant Variables

Some variables do not change by wave and therefore do not contain a wave number.

The RAND convention may use `A` to indicate this type of variable.

Example:

```text
RABYEAR
```

represents respondent birth year.

### Exit Interview Variables

Some variables contain `E` to identify information associated with an Exit Interview.

This distinction is important because not every RAND variable should be interpreted as a repeated wave observation.

# 14. Derived Variables

A major feature of the RAND HRS Longitudinal File is the use of **derived variables**.

RAND does not simply copy raw HRS variables into the Longitudinal File.

RAND processes information from the underlying HRS data to create standardized variables that are more convenient for analysis.

For example, the RAND presentation describes derived measures such as total household income, which may combine multiple underlying income sources.

RAND documentation provides information about how derived variables were constructed.

Documentation includes:

* Variable name
* Variable label
* Variable type
* Descriptive statistics
* Categorical value codes
* Variable construction
* Missing-value information
* Cross-wave differences
* Original HRS variables used in construction

This information is extremely important when developing the Silver CDM.

# 16. Missing Values

RAND HRS variables require careful treatment of missing and non-substantive values.

A missing value does not necessarily mean the same thing as:

* Respondent did not answer
* Question was not asked
* Respondent was not eligible
* Not applicable
* Don't know
* Refused

The RAND documentation describes how variables are constructed and provides information about missing values.

For this reason, Silver CDM DML specifications should explicitly document any transformation of RAND missing-value codes.

# 17. Survey Participation

The RAND HRS Longitudinal File provides participation indicators.

One important variable is:

```text
INWw
```

where `w` represents the wave number.

These variables identify whether a respondent completed a particular survey wave.

This information can be useful when determining whether a respondent has an actual observation for a given wave.

# 18. Spouse Relationships

HRS is not limited to isolated individual records.

Respondents may have spouses or partners who are also represented in the study.

The RAND Longitudinal File includes:

```text
SwHHIDPN
```

which identifies the HHIDPN associated with a respondent's spouse for a given wave.

This provides opportunities to study:

* Spousal health
* Household economics
* Caregiving
* Retirement decisions
* Family relationships
* Joint financial circumstances

# 19. Family Data

The RAND HRS Longitudinal File contains some family-related variables, but it should not be considered a comprehensive family database.

For example, the Longitudinal File contains measures such as:

* Number of people living in the household
* Number of living children
* Number of living brothers and sisters
* Number of children ever born

The RAND HRS Family Data Files provide considerably more detailed information about:

* Children
* Children-in-law
* Parents
* Siblings
* Family assistance
* Financial transfers
* Caregiving

The Family Data Files therefore supplement the Longitudinal File when detailed family analysis is required.

---

# 20. Imputed Data

Some HRS economic and wealth measures require imputation.

The RAND HRS Detailed Imputations File provides additional information about these imputations.

For example, the RAND presentation discusses mortgage and wealth variables and associated imputation and ownership flags.

This is important for researchers because an analytical value may represent an imputed estimate rather than a directly reported value.

Researchers should consult the RAND documentation before interpreting these variables.

# 21. Exit and Post-Exit Data

HRS also collects information after a respondent's death.

An **Exit Interview** is generally conducted with a proxy informant for a deceased respondent.

The proxy may provide information about:

* Health
* Family
* Medical care
* Medical expenditures
* Assets
* Estate distribution
* Family decision-making

RAND also provides Post-Exit Interview data for certain unresolved financial matters.

Some Exit information is incorporated into the RAND HRS Longitudinal File, while the dedicated Exit/Post-Exit products provide additional detail.

# 22. Why the HRS Is Valuable for Public Health

The HRS is particularly valuable for public health because health can be examined together with the social and economic conditions surrounding an individual.

For example:

```text
                    HEALTH
                       │
       ┌───────────────┼───────────────┐
       ▼               ▼               ▼
   Income           Employment       Insurance
       │               │               │
       ▼               ▼               ▼
    Wealth          Retirement       Health Care
       │               │               │
       └───────────────┼───────────────┘
                       ▼
                 Family Structure
                       │
                       ▼
                Longitudinal Change
```

This multidimensional structure allows researchers to examine relationships that are difficult to study using a single-domain health dataset.

# 33. Why the RAND HRS Data Model Matters

The original RAND HRS Longitudinal File is highly useful but remains a wide longitudinal dataset.

The Silver CDM introduces a relational structure that makes the data easier to query.

The transformation is:

```text
SOURCE
──────────────────────────────
HHIDPN
R1AGEY_E
R2AGEY_E
R3AGEY_E
R1MSTAT
R2MSTAT
R3MSTAT
...
          │
          │
          ▼
      Transformation
          │
          ▼
TARGET
──────────────────────────────
respondent_id
wave_id
agey_e
mstat
...
```

This design allows analysts to write queries based on:

```text
Respondent
     +
Wave
     +
Subject
```

rather than repeatedly referencing wave-specific source columns.

# 40. Final Perspective

The most important characteristic of HRS is not simply the number of variables it contains.

Its value comes from the **combination of longitudinal follow-up, multiple cohorts, health information, economic information, employment information, family information, and retirement information**.

RAND enhances this value by providing processed and derived data products that make substantial portions of the HRS easier to analyze.

The RAND HRS Longitudinal File is particularly valuable as a starting point because it brings together information across waves and provides standardized derived variables. It can also be supplemented with other RAND and HRS data products when a research question requires additional detail.

The HRS Silver CDM project builds on this foundation by organizing the information into a relational, respondent-by-wave analytical structure.

The ultimate objective is:

```text
Understand the HRS
        ↓
Understand the RAND HRS
        ↓
Understand the Variables
        ↓
Understand the Cohorts and Waves
        ↓
Build the Silver CDM
        ↓
Perform Longitudinal Analysis
        ↓
Create Visualizations
        ↓
Generate Evidence
        ↓
Support Public Health Policy
```

The project therefore provides a bridge between a complex longitudinal research dataset and practical analytical applications for **hospitals, health systems, state agencies, federal agencies, researchers, and policy analysts**.

The official HRS and RAND documentation should always remain the authoritative sources for interpreting the underlying survey methodology, variables, weighting, missing values, and derived-variable construction.
