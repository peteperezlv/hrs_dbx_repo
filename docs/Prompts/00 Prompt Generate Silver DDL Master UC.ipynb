# Design philosophy

For this HRS Silver CDM, keep the DDL very disciplined:

1. System-generated surrogate PK
BIGINT GENERATED ALWAYS AS IDENTITY

2. Required foreign keys
Only the relationships that were defined.

3. HRS source attributes
Only attributes identified from the HRS Codebook as required for the entity/business requirements.

4. Audit columns
If we have established audit columns as part of your Silver CDM standard, those can remain consistently across all five tables.

5. No invented business attributes
No active, effective dates, status fields, labels, descriptions, etc., unless they are explicitly part of the model/specification.
