# 3. Purpose

### 🟡 MASTER CONTROLLED

This specification defines the requirements for generating a Databricks SQL DDL script for a **Gold-layer analytical table**.

Gold tables contain curated data designed for analytical consumption, including:

* Data analytics
* Business intelligence
* Reporting
* Dashboards
* Data visualization
* Statistical analysis
* Public health analysis
* Policy analysis

The generated DDL must create the physical Gold Delta table required by the analytical use case.

The Gold table should contain only the structural elements required to support the defined analytical purpose.

### DDL Responsibility

The DDL specification defines:

> **What the Gold table looks like.**

It does not define:

> **How data is transformed or loaded into the Gold table.**

# 4. Gold Layer Role

### 🟡 MASTER CONTROLLED

The Gold layer represents the **consumption and analytical layer** of the HRS data architecture.

```text
RAND HRS Dataset
       │
       ▼
Bronze Layer
Raw / Source Data
       │
       ▼
Silver CDM
Standardized Relational Data
       │
       ▼
Gold Layer
Curated Analytical Data
       │
       ├───────────────┐
       ▼               ▼
   Analytics       Visualization
       │               │
       └───────┬───────┘
               ▼
        Policy Insights
```

Gold tables should be designed around **analytical requirements**, rather than simply reproducing Silver CDM tables.

## Table type definitions.
### 🟡 MASTER CONTROLLED

| Table Type     | Overview   |
| -------------- | ---------- |
| **Fact**       | Stores measurable business or analytical events, often at a defined grain such as respondent × wave. Typically contains **foreign keys to dimensions plus measures** such as counts, averages, or totals.                                                            |
| **Dimension**  | Stores descriptive attributes used to **categorize, filter, group, and provide context** for facts. Examples include cohort, wave, race, gender, or geography.                                                                                                       |
| **Aggregate**  | Stores **pre-calculated summaries** of data to improve analytical performance or simplify reporting. Typically contains grouping dimensions plus calculated measures such as count, mean, min, or max.                                                               |
| **Analytical** | A broader category for a table specifically designed to answer an **analytical question or support a particular use case**. It may combine dimensions, measures, and derived attributes and does not necessarily follow a strict star-schema fact/dimension pattern. |

## Gold Table Naming Convention

### 🟡 MASTER CONTROLLED

Gold table names should follow this convention.
| Table Type     | Naming Convention   |
| -------------- | ------------------- |
| **Fact**       | fact_<subject>      |
| **Dimension**  | dim_<subject>       |
| **Aggregate**  | aggregate_<subject> |
| **Analytical** | analytic_<subject>  |

# 18. DDL / DML Separation

### 🟡 MASTER CONTROLLED

The HRS Gold architecture separates physical table creation from data population.

```text
Analytical Use Case
        │
        ▼
Gold DDL Specification
        │
        ▼
Generate DDL
        │
        ▼
Execute DDL
        │
        ▼
Gold Delta Table
        │
        ▼
Validate Structure
        │
        ▼
Gold DML Specification
        │
        ▼
Transform and Load Data
```

### DDL

Defines:

> **What the table looks like.**

### DML

Defines:

> **How the table is populated.**

The two specifications should not duplicate responsibilities.

# 23. What Does NOT Belong in This Specification?

### 🟡 MASTER CONTROLLED

The following information belongs in the **Gold DML Functional Specification**, not the Gold DDL specification:

* Source Silver tables used for loading
* RAND HRS source variables
* Source-to-target mappings
* Source data transformations
* Lookups
* Unpivoting
* Pivoting
* Aggregations
* Calculated values
* Filtering rules
* Business transformation rules
* INSERT logic
* UPDATE logic
* MERGE logic
* Data-loading sequence

The DDL specification should remain focused on the **physical structure of the Gold table**.

# 24. Gold DDL → Gold DML Handoff

### 🟡 MASTER CONTROLLED

The DDL specification is considered complete when the physical Gold table structure has been defined and validated.

The DML specification begins after the DDL structure has been approved.

```text
          GOLD DDL
             │
             ▼
      Physical Table
             │
             ▼
       DDL Validation
             │
             ▼
      APPROVED TABLE
             │
             ▼
          GOLD DML
             │
             ▼
     Source + Mapping
             │
             ▼
       Transformation
             │
             ▼
           Load
```

The Gold DML specification may reference the approved DDL table structure rather than redefining it.

### Step 10 — Validate

Validate the physical table structure.

### Step 11 — Begin DML

Once the DDL is approved, create the corresponding Gold DML specification.

# 26. Architectural Principle

### 🟡 MASTER CONTROLLED

The HRS Gold Layer follows the principle:

> **Define the analytical requirement first, define the physical table second, and define the data-loading process third.**

This produces a controlled separation:

```text
Business Requirement
        │
        ▼
Analytical Design
        │
        ▼
Gold DDL
"What does it look like?"
        │
        ▼
Gold Table
        │
        ▼
Gold DML
"How is it populated?"
```

The result is a reusable, maintainable, and consistent framework for expanding the HRS Gold analytical layer across subject areas such as:

* Demographics
* Health
* Financial
* Employment

* Retirement
* Family
* Housing
* Social Security
* Other HRS analytical domains
