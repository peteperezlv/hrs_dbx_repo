The 'Wave' column values are strings, not integers and should be handled appropriately in the DDL and DML SQL code.

### Business Key Uniqueness Constraint
The generated DDL must enforce:

Databricks Runtime 15.x supports:

- PRIMARY KEY
- FOREIGN KEY
- UNIQUE
- CHECK
- NOT NULL
- GENERATED ALWAYS AS IDENTITY

The specification must explicitly require:

| Constraint Type | Required | Notes |
| --- | --- | --- |
| PRIMARY KEY | ✓ | Enforced at write time |
| FOREIGN KEY | ✓ | Enforced at write time |
| UNIQUE | ✓ | Required for business key |
| NOT NULL | ✓ | Required for audit + FK columns |
| CHECK | Optional | Not required for health |
| IDENTITY | ✓ | Required for surrogate key |
