# Research extracted text

Use this phython script for when you want to search for and confirm the existance of a TERM within the pages table.

```python
# ------------------------------------------------------------
# Configuration — extracted Codebook pages table.
# Defined here so this notebook is self-contained and does not
# depend on Step 1.2's configuration cell being run first.
# ------------------------------------------------------------

PAGES_TABLE = "dev_catalog.hrs_agent_metadata.hrs_document_pages"
```

```python
# Cell 8 — Search the extracted text (example)

# Replace the search term to test a variable, concept, or phrase.
SEARCH_TERM = "R1BMI"

from pyspark.sql import functions as F

display(
    spark.table(PAGES_TABLE)
         .filter(F.col("page_text").contains(SEARCH_TERM))
         .select("page_number", "review_flag", "page_text")   # If the review flag is true, then this page needs OCR or visual review!
         .orderBy("page_number")
         .limit(20)
)
```
