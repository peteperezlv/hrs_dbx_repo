Purpose: partition the HRS raw table into partition to improve query performance.

Benefits.
1. Replaces traditional table partitioning and z-ordering.
2. Combines smaller files and dlivers larger files to enhance query performance.
3. It is incremental - data is only rewritten when necessary.
4. Provides flexibility to change clustering keys without rewriting existing data.
5. Beneficial for columns with high cardinaltity unlike partitioning.

Questions Before Using Liquid Clustering
1. How big is the table, and what is its write parttern?
    Equid Clustering makes sense for large tables and lots of activity for the engine to learn from.
2. What is the query shaped.  If the querys shift with new columns in the queries as the business evolves, then Liquid Clustering.
    If they are simple and consistent and the table is large, then partition.
3. What are the retention and compliance (regulatory) constraints.  If the data is deleted by some date, then partition.
4. What does it cost to migrate from Partitioning to Liquid Clustering.
5. Partition when
    a) Regulatory Delete patterns
    b) Rention driven tables.  Droping partitions to remove stale data are cheaper.
    c) Very small tables like dimension tables.
    d) Access patterns that Liquid Clustering does not server well like: queries that require full table scans like "Select all cohots across all waves"
