## Purpose
The purpose of this document is to serve as an interactive learning resource and technical reference for data analysts, researchers, and SQL users querying the RAND Health and Retirement Study (HRS) relational database in Databricks. It establishes a standardized framework for understanding how longitudinal survey data is structured and accessed within our data platform.

## Objective
The objective of these sample queries is to provide a practical, hands-on learning guide for navigating the relational structure of the database. By transitioning through increasing levels of complexity—from basic data retrieval to advanced analytical techniques like multi-table joins, window functions, and Common Table Expressions (CTEs)—these examples demonstrate how to efficiently extract, aggregate, and analyze survey responses across various waves, cohorts, and individual respondents.

## Understanding Our HRS Data Structure
To make the huge amount of HRS survey information easier to use, we have organized it like a highly organized digital filing cabinet in Databricks. Instead of keeping everything in one giant, messy spreadsheet, we have separated the information into six distinct tables (which you can think of as organized lists).

### Here are the six lists available to you:

1. HRS Cohort: A master list of the different age groups (cohorts) participating in the study.

1. HRS Wave: A master list of the specific years or cycles (waves) when the interviews took place.

1. HRS Respondent: A master list of the individual people who answered the survey.

1. HRS Demographics: Background information about the people (such as age, gender, and marital status).

1. HRS Health: Information regarding the physical and mental health responses of the participants.

1. HRS Leave-Behind: Special data from paper surveys that participants filled out and mailed back later.

### How the Lists Connect (The Relationships)
To get a complete picture of a person's answers, you will often need to connect these lists. They link together using simple, logical rules:

* **One Group, Many People**: Each **Cohort** (age group) contains many different **Respondents** (individual people).

* **One Person, Many Records**: Over time, a single **Respondent** will have multiple entries in the **Demographics**, **Health**, and **Leave-Behind** lists because they participate in the survey year after year.

* **One Survey Year, Many Records**: Each **Wave** (survey year) contains answers from thousands of different people, meaning a single wave connects to many rows in the **Demographics, Health, and Leave-Behind** lists.
