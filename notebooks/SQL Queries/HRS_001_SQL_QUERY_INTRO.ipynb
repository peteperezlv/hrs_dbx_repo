## What is SQL
SQL (pronounced "Sequel" or "S-Q-L") stands for Structured Query Language.

It is the standard digital language used to communicate with databases. Think of it as a universal remote control for data: instead of clicking through endless spreadsheets, you write short text commands to talk directly to your data tables.

## What Does SQL Do?
In a data platform like Databricks, you use SQL to perform four main actions (often called CRUD):
* Create: Add new tables or input new survey answers.
* Read: Search, filter, and extract specific information (e.g., "Find all respondents over age 65").
* Update: Change or fix existing information when new data arrives.
* Delete: Remove outdated or incorrect records.

### An Everyday Analogy
Imagine a massive library containing millions of books.
* Without SQL, you would have to walk down every aisle, pulling books off shelves one by one to find what you want.
* With SQL, you hand a specific, written request to a magical librarian (like Databricks) who brings you the exact pages you need in a fraction of a second.

## The Big Six of SQL
Think of a SQL query as a set of step-by-step instructions given to a librarian to fetch and organize information from a filing cabinet. To get the correct results, you must write these instructions in a specific order.

Here is the basic anatomy of a SQL query, explained in the exact order you must write it:

* SELECT (What you want to see?)

    * This is where you list the specific columns or pieces of information you want to display in your final results.
    * The Analogy: Telling the librarian, "Show me only the names and phone numbers."
    * Example: SELECT respondent_id, age

* FROM (Where is the data stored?)
    * This tells the database which table (or list) contains the columns you just asked for in the SELECT step.
    * The Analogy: Telling the librarian, "Look inside the Demographics drawer."
    * Example: FROM hrs_demographics

* WHERE (How do you want to filter individual records?)
    * This filters out rows that do not meet your specific conditions before any calculations happen. It narrows down your search.
    * The Analogy: Telling the librarian, "Only look at people who are older than 65."
    * Example: WHERE age > 65

* GROUP BY (How do you want to slice the data?)
    * This collapses individual rows into summary categories so you can calculate totals, averages, or counts for groups of data.
    * The Analogy: Telling the librarian, "Separate these files into distinct stacks based on gender."
    * Example: GROUP BY gender

* HAVING (How do you want to filter your summary categories?)
    * This acts exactly like a WHERE clause, but it is used exclusively to filter the grouped categories created in the GROUP BY step. It filters after calculations happen.
    * The Analogy: Telling the librarian, "Now, throw away any gender stack that has fewer than 100 files in it."
    * Example: HAVING COUNT(respondent_id) > 100

* ORDER BY (How should the final list be sorted?)
    * This sorts your final output so it is easy to read. You can sort numbers from lowest to highest, or text alphabetically.
    * The Analogy: Telling the librarian, "Alphabetize the final list by the person's last name before handing it to me."
    * Example: ORDER BY age DESC (DESC means highest to lowest)
