## How This Simulates DAB Variable Orchestration

This notebook demonstrates how to:
1. **Simulate DAB variables** at the orchestration level
2. **Pass variables to child notebooks** using `dbutils.notebook.run()`
3. **Orchestrate multiple notebooks** in sequence

### What Happens

* Variables `catalog_name` and `schema_prefix` are defined here
* They are passed to the child notebook via the `arguments` parameter
* The child notebook must accept these as widgets/parameters to use them
