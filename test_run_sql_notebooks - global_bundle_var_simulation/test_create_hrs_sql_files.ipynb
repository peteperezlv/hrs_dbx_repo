## How This Simulates DAB Variables

This notebook simulates how Declarative Automation Bundle (DAB) variable interpolation works:

1. **Variables defined**: `catalog_name = "dev_catalog"` and `schema_prefix = "slv_cdm_hrs"`
2. **SQL file read**: Reads `/sql/ddl/create_hrs_cohort.sql`
3. **String replacement**: Replaces hardcoded values with the variables (simulating `${catalog_name}` interpolation)
4. **Execution**: Splits multiple SQL statements and executes each

### For Production DAB Deployment:

In your `databricks.yml`, define variables:
```yaml
variables:
  catalog_name:
    default: dev_catalog
  schema_prefix:
    default: slv_cdm_hrs
```

Then reference them in SQL as `${catalog_name}` and `${schema_prefix}`. The bundle will interpolate these at deployment time.
